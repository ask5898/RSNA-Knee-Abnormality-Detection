# RSNA knee: submission from the `rsna_knee` package

The short way to submit with this repo on Kaggle (Internet off). Attach three inputs:

1. the competition data,
2. this repository, uploaded as a Kaggle dataset (any name; the cell below finds `src/rsna_knee`),
3. a trained `knee_model/` folder (`backbone.pt`, `head_fold*.pt`, `config.json`) as a dataset.

Training happens elsewhere: `scripts/train.py`, or the full notebook `rsna-knee-abnormality.ipynb` with `TRAIN = True`.

```python
import os
import sys

INPUT = "/kaggle/input"


def find(test, root=INPUT):
    """First folder under root for which test(folder, its file names) holds; skips the DICOM folders."""
    for folder, dirs, files in os.walk(root):
        dirs[:] = [d for d in dirs if d != "competitions"]
        if test(folder, set(files)):
            return folder
    raise FileNotFoundError(f"nothing matching under {root}")


package = find(lambda folder, files: os.path.basename(folder) == "rsna_knee" and "config.py" in files)
sys.path.insert(0, os.path.dirname(package))

from rsna_knee import config as C

C.DATA_PATH = os.path.join(INPUT, "competitions", "rsna-knee-abnormality-detection")
C.MODEL_DIR = find(lambda folder, files: {"config.json", "backbone.pt"} <= files)
print("package:", package, "| model:", C.MODEL_DIR)
```
